```python

df['hours_logged'] = pd.to_numeric(df['hours_logged'], errors='coerce')
```

```python
df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'].str.replace('$', '', regex=False), errors='coerce')
```

```python
df['csat_score'] = pd.to_numeric(df['csat_score'], errors='coerce')
```

```python
print(df['hours_logged'].mean(), df['hours_logged'].min(), df['hours_logged'].max())
```

```python
df.sort_values('hours_logged', ascending=False).head(10)
```

```python
df.sort_values('billable_amount_usd', ascending=False).head(5)
```

```python
df['client_name'] = df['client_name'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

```python
df['client_name'] = df['client_name'].str.title()
```

```python
df['ticket_created_at'] = pd.to_datetime(df['ticket_created_at'], errors='coerce', format='mixed')
```

```python
df['ticket_year'] = df['ticket_created_at'].dt.year
```

```python
df['ticket_month'] = df['ticket_created_at'].dt.month
```

```python
df['ticket_year'].value_counts().sort_index()
```

```python
df['total'] = df['billing_rate_(usd)'] * df['hours_logged']
```

```python
df['csat_score'] = df['csat_score'].fillna(df['csat_score'].mean())
```

```python
df.dropna(subset=['client_name'], inplace=True)
```
